# Taller de Condicionales y Bucles

En este taller se practican las estructuras de control de flujo de Python mediante el uso de `if`, `elif`, `else`, `for`, `range()`, `while`, `break` y `continue`.

## Ejercicio 1 — Clasificar el stock de un producto

En este ejercicio se clasifica la cantidad disponible de un producto en inventario. Se utiliza una estructura `if`, `elif` y `else` para determinar si el producto está agotado, tiene stock bajo o cuenta con stock suficiente.

```python
cantidad_en_stock = 3

if cantidad_en_stock == 0:
    nivel = "Agotado"
elif cantidad_en_stock < 5:
    nivel = "Stock bajo"
else:
    nivel = "Stock suficiente"

print("Nivel de stock:", nivel)
```

## Ejercicio 2 — Alerta de reposición inmediata

En este ejercicio se determina si es necesario reponer el producto. Se utiliza el nivel de stock obtenido anteriormente y la cantidad disponible, combinando ambas condiciones mediante el operador lógico `or`.

```python
if nivel == "Agotado" or cantidad_en_stock < 5:
    print("Reponer pronto")
else:
    print("Sin acción requerida")
```

## Ejercicio 3 — Clasificar el clima del lunes

En este ejercicio se clasifica la temperatura registrada el lunes como fría, templada o calurosa. Se utiliza una estructura `if`, `elif` y `else` para cubrir los tres posibles rangos de temperatura.

```python
temp_lunes = 18.5

if temp_lunes < 15:
    clima = "Frío"
elif temp_lunes >= 15 and temp_lunes <= 25:
    clima = "Templado"
else:
    clima = "Caluroso"

print("Clima del lunes:", clima)
```

## Ejercicio 4 — Validar el consumo de la factura

En este ejercicio se verifica si el consumo de energía eléctrica está dentro del rango considerado normal por el proveedor. Se utilizan dos comparaciones combinadas con `and`, incluyendo los límites inferior y superior del rango.

```python
consumo_kwh = 142
tarifa_kwh = 623.5

if consumo_kwh >= 100 and consumo_kwh <= 200:
    print("Consumo normal")
else:
    print("Consumo fuera de rango, revisar con el proveedor")
```

## Ejercicio 5 — Aprobación con mensaje

En este ejercicio se calculan las tres notas parciales de un estudiante y su promedio. Luego se clasifica el resultado mediante `if`, `elif` y `else` en reprobado, aprobado o aprobado con excelencia, según el promedio obtenido.

```python
nota_1 = 4.2
nota_2 = 3.5
nota_3 = 4.8

promedio = (nota_1 + nota_2 + nota_3) / 3

if promedio < 3.0:
    resultado = "Reprobado"
elif promedio >= 3.0 and promedio <= 4.4:
    resultado = "Aprobado"
else:
    resultado = "Aprobado con excelencia"

print("Promedio:", promedio)
print("Resultado:", resultado)
```

## Ejercicio 6 — Recorrer el inventario completo

En este ejercicio se recorre una lista que contiene las cantidades disponibles de varios productos. Se utiliza `for` junto con `range(len(...))` para acceder a cada posición de la lista y clasificar cada cantidad como agotada, stock bajo o stock suficiente.

```python
cantidades_en_stock = [0, 3, 12, 0, 7, 2]

for indice in range(len(cantidades_en_stock)):
    cantidad = cantidades_en_stock[indice]

    if cantidad == 0:
        nivel = "Agotado"
    elif cantidad < 5:
        nivel = "Stock bajo"
    else:
        nivel = "Stock suficiente"

    print(f"Producto {indice}: {nivel}")
```

## Ejercicio 7 — Promedio semanal de temperaturas

En este ejercicio se recorren las temperaturas registradas durante la semana utilizando un acumulador para sumar todas las temperaturas y un contador para determinar cuántos días superaron los 20 grados. Al finalizar el recorrido se calcula el promedio semanal.

```python
temperaturas_semana = [18.5, 21.0, 19.8, 24.3, 17.2]

suma_temperaturas = 0
dias_calurosos = 0

for indice in range(len(temperaturas_semana)):
    temperatura = temperaturas_semana[indice]

    suma_temperaturas = suma_temperaturas + temperatura

    if temperatura > 20:
        dias_calurosos = dias_calurosos + 1

promedio_semanal = suma_temperaturas / len(temperaturas_semana)

print("Promedio semanal:", promedio_semanal)
print("Cantidad de días calurosos:", dias_calurosos)
```

## Ejercicio 8 — Comprar tornillos mientras alcance el presupuesto

En este ejercicio se simula la compra de tornillos mientras el presupuesto disponible sea suficiente. Se utiliza un bucle `while`, descontando el precio de cada tornillo del presupuesto e incrementando un contador de unidades compradas en cada repetición.

```python
presupuesto_ferreteria = 3000
precio_tornillo = 250

tornillos_comprados = 0

while presupuesto_ferreteria >= precio_tornillo:
    presupuesto_ferreteria = presupuesto_ferreteria - precio_tornillo
    tornillos_comprados = tornillos_comprados + 1

print("Tornillos comprados:", tornillos_comprados)
print("Presupuesto sin usar:", presupuesto_ferreteria)
```

## Ejercicio 9 — Buscar la primera herramienta cara

En este ejercicio se recorren directamente los precios de las herramientas para encontrar el primer precio superior a 20.000 pesos. Mientras no se encuentre un precio que supere ese valor, se muestra un mensaje indicando que la revisión continúa. Al encontrarlo, se muestra el precio y se utiliza `break` para detener el bucle inmediatamente.

```python
precios_herramientas = [12000.0, 8500.0, 25000.0, 9800.0]

for precio in precios_herramientas:
    if precio > 20000:
        print(f"Primera herramienta cara encontrada: ${precio}")
        break
    else:
        print(f"Sigue revisando: ${precio}")
```

## Ejercicio 10 — Revisión completa del inventario de la ferretería

En este ejercicio se revisa un inventario de herramientas utilizando dos listas relacionadas por posición. Se clasifica la cantidad disponible mediante `if`, `elif` y `else`. Si una herramienta tiene stock suficiente, se utiliza `continue` para pasar a la siguiente. Si está agotada, se muestra una alerta y se utiliza `break` para detener la revisión. Para stock bajo se muestra la información y el recorrido continúa normalmente.

```python
nombres_herramientas = [
    "Martillo",
    "Taladro",
    "Sierra",
    "Destornillador",
    "Nivel",
    "Llave inglesa"
]

cantidades_herramientas = [5, 0, 2, 8, 0, 1]

for indice in range(len(nombres_herramientas)):
    nombre = nombres_herramientas[indice]
    cantidad = cantidades_herramientas[indice]

    if cantidad == 0:
        nivel = "Agotado"
    elif cantidad < 5:
        nivel = "Stock bajo"
    else:
        nivel = "Stock suficiente"

    if nivel == "Stock suficiente":
        continue

    elif nivel == "Agotado":
        print(f"Alerta: reponer urgentemente {nombre}.")
        break

    else:
        print(f"Stock bajo: {nombre}, cantidad disponible: {cantidad}")
```
