Se declaran los metadatos de la estación, usando el tipo de dato solicitado para cada variable. Luego se utiliza type() para verificar los tipos de las siete variables

```python
codigo_estacion = "EST-01"
sector = "Norte"
altitud_m = 1495
numero_lecturas = 305
latitud = 6.28
sensor_calibrado = True
factor_conversion = 1000

print("Tipo de codigo_estacion:", type(codigo_estacion))
print("Tipo de sector:", type(sector))
print("Tipo de altitud_m:", type(altitud_m))
print("Tipo de numero_lecturas:", type(numero_lecturas))
print("Tipo de latitud:", type(latitud))
print("Tipo de sensor_calibrado:", type(sensor_calibrado))
print("Tipo de factor_conversion:", type(factor_conversion))
```

Ejercicio 2 — Conversión de unidades y error frente al equipo patrón
Se convierte la lectura del sensor de mg/m³ a µg/m³ usando el factor definido en el ejercicio anterior. Después se calculan el error absoluto con signo, el error relativo porcentual y el error cuadrático mediante operadores aritméticos.

```python
lectura_sensor_mg = 0.0418
valor_patron_ug = 40.0

lectura_sensor_ug = lectura_sensor_mg * factor_conversion
error_absoluto = lectura_sensor_ug - valor_patron_ug
error_relativo_porcentual = (error_absoluto / valor_patron_ug) * 100
error_cuadratico = error_absoluto ** 2

print("Lectura del sensor convertida (µg/m³):", lectura_sensor_ug)
print("Error absoluto (µg/m³):", error_absoluto)
print("Error relativo porcentual (%):", error_relativo_porcentual)
print("Error cuadrático:", error_cuadratico)
```

Ejercicio 3 — Cobertura temporal del lote descargado
Se calcula la cobertura temporal de las lecturas descargadas. La división entera permite obtener horas y días completos, mientras que el módulo determina las lecturas y horas sobrantes.

```python
lecturas_por_hora = 12

horas_completas = numero_lecturas // lecturas_por_hora
lecturas_sobrantes = numero_lecturas % lecturas_por_hora
dias_completos = horas_completas // 24
horas_sobrantes = horas_completas % 24

print("Horas completas:", horas_completas)
print("Lecturas sobrantes:", lecturas_sobrantes)
print("Días completos:", dias_completos)
print("Horas sobrantes:", horas_sobrantes)
```

Ejercicio 4 — Validar una lectura sin decidir todavía
Se comprueba si una lectura corresponde al centinela de dato faltante, si está dentro del rango físico del sensor y si finalmente es válida. En este ejercicio solo se evalúan expresiones booleanas; no se usa if.

```python
lectura = 41.8
codigo_dato_faltante = -999.0
limite_superior_sensor = 500.0

es_dato_faltante = lectura == codigo_dato_faltante
esta_en_rango = lectura >= 0 and lectura <= limite_superior_sensor
lectura_valida = esta_en_rango and not es_dato_faltante

print("¿Es dato faltante?:", es_dato_faltante)
print("¿Está en rango?:", esta_en_rango)
print("¿La lectura es válida?:", lectura_valida)
```

Se clasifica la lectura en una de las cuatro categorías oficiales de PM2.5 mediante una sola cadena if / elif / else, con umbrales ordenados de menor a mayor. Los límites se definen sin solapamientos: 12.0 pertenece a Buena, 35.4 a Moderada y 55.4 a Dañina para grupos sensibles.

Ejercicio 6 — Limpiar la serie horaria
Se recorren las doce lecturas. Cuando aparece el centinela -999.0, se incrementa el contador de datos descartados y continue salta a la siguiente iteración. Las lecturas válidas se guardan en una lista y se acumulan para calcular el promedio al terminar el bucle.

```python
lecturas_pm25 = [8.4, 12.7, -999.0, 15.2, 41.8, 9.6, -999.0, 58.3, 22.0, 13.9, 7.5, 36.1]
lecturas_validas = []
suma_pm25 = 0.0
lecturas_descartadas = 0

for valor_lectura in lecturas_pm25:
    if valor_lectura == -999.0:
        lecturas_descartadas = lecturas_descartadas + 1
        continue

    lecturas_validas.append(valor_lectura)
    suma_pm25 = suma_pm25 + valor_lectura

promedio_pm25 = suma_pm25 / len(lecturas_validas)

print("Cantidad de lecturas válidas:", len(lecturas_validas))
print("Cantidad de lecturas descartadas:", lecturas_descartadas)
print("Promedio PM2.5 (µg/m³):", promedio_pm25)
```

Ejercicio 7 — Estadística descriptiva de la serie limpia
Se calculan el máximo y el mínimo recorriendo la lista con un bucle y comparaciones, sin usar max(), min() ni sum(). Luego, en un segundo recorrido, se suman las desviaciones cuadráticas respecto al promedio y se calcula la desviación estándar poblacional usando la potencia ** 0.5.

```python
maximo_pm25 = lecturas_validas[0]
minimo_pm25 = lecturas_validas[0]

for valor_lectura in lecturas_validas:
    if valor_lectura > maximo_pm25:
        maximo_pm25 = valor_lectura
    if valor_lectura < minimo_pm25:
        minimo_pm25 = valor_lectura

rango_pm25 = maximo_pm25 - minimo_pm25

suma_desviaciones = 0.0
for valor_lectura in lecturas_validas:
    suma_desviaciones = suma_desviaciones + (valor_lectura - promedio_pm25) ** 2

desviacion_estandar = (suma_desviaciones / len(lecturas_validas)) ** 0.5

print("Máximo PM2.5:", maximo_pm25)
print("Mínimo PM2.5:", minimo_pm25)
print("Rango PM2.5:", rango_pm25)
print("Desviación estándar poblacional:", desviacion_estandar)
```

Ejercicio 8 — Decaimiento tras cerrar la fuente de emisión
Se simula la disminución de la concentración a razón del 15 % por hora. El bucle while continúa mientras la concentración sea superior a 12.0 µg/m³; en cada iteración se actualiza la concentración y se incrementa el contador de horas.

```python
concentracion = 58.3
factor_decaimiento = 0.85
umbral_buena = 12.0
horas_transcurridas = 0

while concentracion > umbral_buena:
    concentracion = concentracion * factor_decaimiento
    horas_transcurridas = horas_transcurridas + 1

print("Horas transcurridas:", horas_transcurridas)
print("Concentración final (µg/m³):", concentracion)
```

Ejercicio 9 — Registros de varias estaciones
Se representan las estaciones mediante una lista de diccionarios. Para cada estación se desempaquetan las coordenadas de la tupla, se calcula el promedio de sus lecturas con un acumulador y un bucle interno, se obtiene la humedad con get() y un valor por defecto, y se clasifica el promedio usando la escala del taller.

```python
registros_estaciones = [
    {"codigo": "EST-01", "sector": "Norte", "coordenadas": (6.28, -75.57), "lecturas": [8.4, 12.7, 15.2]},
    {"codigo": "EST-02", "sector": "Centro", "coordenadas": (6.24, -75.58), "lecturas": [41.8, 58.3, 36.1]},
    {"codigo": "EST-03", "sector": "Sur", "coordenadas": (6.19, -75.59), "lecturas": [9.6, 13.9, 7.5], "humedad_relativa": 68.0},
]

for registro in registros_estaciones:
    latitud, longitud = registro["coordenadas"]

    suma_lecturas_estacion = 0.0
    for valor_lectura in registro["lecturas"]:
        suma_lecturas_estacion = suma_lecturas_estacion + valor_lectura

    promedio_estacion = suma_lecturas_estacion / len(registro["lecturas"])
    humedad = registro.get("humedad_relativa", "no disponible")

    if promedio_estacion <= 12.0:
        categoria_estacion = "Buena"
    elif promedio_estacion <= 35.4:
        categoria_estacion = "Moderada"
    elif promedio_estacion <= 55.4:
        categoria_estacion = "Dañina para grupos sensibles"
    else:
        categoria_estacion = "Dañina para la salud"

    print(
        "Código:", registro["codigo"],
        "| Sector:", registro["sector"],
        "| Latitud:", latitud,
        "| Longitud:", longitud,
        "| Promedio:", promedio_estacion,
        "| Categoría:", categoria_estacion,
        "| Humedad:", humedad
    )
```
