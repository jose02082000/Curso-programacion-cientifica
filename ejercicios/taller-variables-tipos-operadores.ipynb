## Ejercicio 1 — Inventario de la tienda

En este ejercicio se declaran las variables relacionadas con el inventario de la tienda, incluyendo el nombre de un producto, los precios, las cantidades a comprar y si existen descuentos activos. También se verifica el tipo de dato de cada variable.

```python
nombre_producto = "Cuaderno"

precio_cuaderno = 3500
precio_lapiz = 800
precio_cafe = 6200.50

hay_descuento = False

cantidad_cuadernos = 3
cantidad_lapices = 5

print(type(nombre_producto))
print(type(precio_cuaderno))
print(type(precio_lapiz))
print(type(precio_cafe))
print(type(hay_descuento))
print(type(cantidad_cuadernos))
print(type(cantidad_lapices))
```

## Ejercicio 2 — Total de la compra y descuento

En este ejercicio se calculan los valores correspondientes a la compra de cuadernos y lápices. Luego se obtiene el total general, se calcula un descuento del 10% y finalmente se obtiene el total que se pagaría después de aplicar el descuento.

```python
total_cuadernos = precio_cuaderno * cantidad_cuadernos
print("Total de cuadernos:", total_cuadernos)

total_lapices = precio_lapiz * cantidad_lapices
print("Total de lápices:", total_lapices)

total_general = total_cuadernos + total_lapices + precio_cafe
print("Total general:", total_general)

descuento = total_general * 0.10
print("Descuento:", descuento)

total_con_descuento = total_general - descuento
print("Total con descuento:", total_con_descuento)
```

## Ejercicio 5 — Promedio y comparación de temperaturas

En este ejercicio se registran las temperaturas de lunes, martes y miércoles y se calcula su promedio. También se calcula la diferencia entre martes y lunes y se realizan varias comparaciones con la temperatura promedio de referencia para determinar diferentes condiciones.

```python
temp_lunes = 18.5
temp_martes = 21.0
temp_miercoles = 19.8
temp_promedio_semana = 20.0

promedio_temperaturas = (temp_lunes + temp_martes + temp_miercoles) / 3
print("Promedio de temperaturas:", promedio_temperaturas)

diferencia_martes_lunes = temp_martes - temp_lunes
print("Diferencia entre martes y lunes:", diferencia_martes_lunes)

print(
    "¿El martes fue más caluroso que el promedio de referencia?",
    temp_martes > temp_promedio_semana
)

print(
    "¿El miércoles fue igual al promedio de referencia?",
    temp_miercoles == temp_promedio_semana
)

print(
    "¿El lunes fue el día más frío?",
    temp_lunes < temp_martes and temp_lunes < temp_miercoles
)
```

## Ejercicio 7 — Factura de servicios y consumo normal

En este ejercicio se calcula el costo del consumo de energía eléctrica y se determina si el consumo se encuentra dentro del rango considerado normal por el proveedor, es decir, entre 100 y 200 kWh.

```python
consumo_kwh = 142
tarifa_kwh = 623.5

total_servicio = consumo_kwh * tarifa_kwh
print("Total a pagar:", total_servicio)

consumo_mayor_igual_100 = consumo_kwh >= 100
print("¿El consumo es mayor o igual a 100 kWh?", consumo_mayor_igual_100)

consumo_menor_igual_200 = consumo_kwh <= 200
print("¿El consumo es menor o igual a 200 kWh?", consumo_menor_igual_200)

consumo_normal = consumo_mayor_igual_100 and consumo_menor_igual_200
print("¿El consumo es normal?", consumo_normal)
```

## Ejercicio 10 — Intercambiar dos precios

En este ejercicio se corrigen los precios de dos repuestos que fueron registrados de manera invertida. Para realizar el intercambio se utiliza una variable temporal, evitando asignar directamente los valores finales a cada variable.

```python
precio_tornillo = 250
precio_tuerca = 180

print("Antes del intercambio:")
print("Precio del tornillo:", precio_tornillo)
print("Precio de la tuerca:", precio_tuerca)

temporal = precio_tornillo
precio_tornillo = precio_tuerca
precio_tuerca = temporal

print("Después del intercambio:")
print("Precio del tornillo:", precio_tornillo)
print("Precio de la tuerca:", precio_tuerca)
```
